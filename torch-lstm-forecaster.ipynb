# Forecast with an LSTM

*Medium · from [mlcode](https://mlcode.tech/problems/torch-lstm-forecaster)*

Train an LSTM that forecasts the next value of a seasonal series.

You get one training series, a 1D tensor. It has a daily and a weekly cycle
on top of a level near 50, plus noise. Write
`train_forecaster(series, window)`. It must:

1. normalise the series with the mean and standard deviation of the
   **training** series,
2. cut it into windows of `window` values, each with the value that follows
   it as the target (`Tensor.unfold` does this in one call),
3. train an `nn.LSTM` with a linear head on the last hidden state,
   using mini-batches from a `DataLoader`,
4. return a `predict(windows)` function. It takes a `(B, window)` tensor of
   raw values and returns a `(B,)` tensor of forecasts in raw units.

### Function signature

```python
def train_forecaster(series, window):
```

### Example

**Input**

```
series = torch.tensor([50.1, 54.3, 57.9, ...])     # 1000 values
predict = train_forecaster(series, window=24)
predict(torch.stack([series[-24:]]))
```

**Output**

```
tensor([53.8])     # one forecast for the step after the window
```

**Reasoning**

The forecast for a window is the model output for its last hidden state,
turned back into raw units with `* std + mean`. The cycles repeat, so the
shape of the last 24 values tells the model where in each cycle the next
step falls.

### Constraints and notes

- The tests forecast windows from the 400 steps after the training series
  (data the model never saw) and need a mean squared error below `1.0` and
  below 20% of the "repeat the last value" baseline, which is about `8.7`.
- Without normalisation, a level of 50 with a small initial network gives
  large gradients and a slow start. Use the training statistics only.
- `nn.LSTM(1, hidden, batch_first=True)` expects `(B, T, 1)`. Add the
  feature axis with `unsqueeze(-1)`.
- Return predictions under `torch.no_grad()` and keep training to a few
  seconds.
